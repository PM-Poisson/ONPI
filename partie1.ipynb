2) Généralisation au cas N-D

Généraliser le code au cas d’un signal x ∈ R^N , permettant de résoudre le problème d’optimisation
suivant :
ˆx = argmin ∥x∥^2
      x∈RN

```python
import numpy as np
import matplotlib.pyplot as plt

def descente_gradient(grad_f, x0, tau, n_iter):
    """
    Algorithme de descente de gradient

    Paramètres
    ----------
    grad_f : fonction
        Gradient de la fonction de coût f
    x0 : float ou np.ndarray
        Initialisation
    tau : float
        Pas de descente
    n_iter : int
        Nombre d'itérations

    Retours
    -------
    x_hist : list
        Historique des itérés
    """
    x = x0
    x_hist = [x0]

    for k in range(n_iter):
        x = x - tau * grad_f(x)
        x_hist.append(x)

    return x_hist
```
